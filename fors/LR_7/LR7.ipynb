```python
# Загрузка данных
data_xx = pd.read_csv('4.3.2 XX.csv', skiprows=2, delim_whitespace=True, 
                     header=None, names=['freq', 'S11', 'angle'])
data_k3 = pd.read_csv('4.3.2 K3.csv', skiprows=2, delim_whitespace=True, 
                     header=None, names=['freq', 'S11', 'angle'])

# Добавляем преобразование в децибелы
data_xx['S11_dB'] = 20 * np.log10(data_xx['S11'])
data_k3['S11_dB'] = 20 * np.log10(data_k3['S11'])

# --- График модуля S11 в децибелах ---
plt.figure(figsize=(10, 6))

# Основные кривые с преобразованными данными
line_xx, = plt.plot(data_xx['freq'], data_xx['S11_dB'], 
                   label='|S11|, XX', color='black', linewidth=2.5)
line_k3, = plt.plot(data_k3['freq'], data_k3['S11_dB'], 
                   label='|S11|, K3', color='orange', linestyle='--', linewidth=2.5)

# Точки на 10 МГц с новыми значениями
point_xx = plt.scatter(10, data_xx['S11_dB'].iloc[0], color='black', s=80, 
                      zorder=5, label=f'XX: {data_xx["S11_dB"].iloc[0]:.2f} dB')
point_k3 = plt.scatter(10, data_k3['S11_dB'].iloc[0], color='orange', s=80, 
                      zorder=5, label=f'K3: {data_k3["S11_dB"].iloc[0]:.2f} dB')

# Обновленные подписи осей
plt.xlabel('Частота, МГц', fontsize=12)
plt.ylabel('Коэффициент отражения, дБ', fontsize=12)  # Изменена единица измерения
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(handles=[line_xx, line_k3, point_xx, point_k3])
plt.tight_layout()
plt.show()

# --- График аргумента S11 (без изменений) ---
plt.figure(figsize=(10, 6))

line_xx_ang, = plt.plot(data_xx['freq'], data_xx['angle'], 
                       label='arg(S11), XX', color='g', linewidth=2.5)
line_k3_ang, = plt.plot(data_k3['freq'], data_k3['angle'], 
                       label='arg(S11), K3', color='m', linestyle='--', linewidth=2.5)

point_xx_ang = plt.scatter(10, data_xx['angle'].iloc[0], color='g', s=80, 
                          zorder=5, label=f'XX: {data_xx["angle"].iloc[0]:.2f}°')
point_k3_ang = plt.scatter(10, data_k3['angle'].iloc[0], color='m', s=80, 
                          zorder=5, label=f'K3: {data_k3["angle"].iloc[0]:.2f}°')

plt.xlabel('Частота, МГц', fontsize=12)
plt.ylabel('Аргумент коэффициента отражения, градусы', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(handles=[line_xx_ang, line_k3_ang, point_xx_ang, point_k3_ang])
plt.tight_layout()
plt.show()
```

```python
# Загрузка данных
data = pd.read_csv('4.4.3.csv', skiprows=2, delim_whitespace=True, 
                  header=None, names=['freq', 'S11', 'angle'])

# Преобразование S11 в децибелы
data['S11_dB'] = 20 * np.log10(data['S11'])

# Выбранные частоты и их индексы в DataFrame
selected_freqs = [10.0, 30.202020, 50.404040]
indices = [0, 1, 2]  # Соответствуют первым трем строкам данных

# --- График модуля S11 в децибелах ---
plt.figure(figsize=(12, 6))

# Основная кривая
line_dB, = plt.plot(data['freq'], data['S11_dB'], 
                  label='|S11| (дБ)', color='black', linewidth=2)

# Точки на выбранных частотах
markers = ['o', 's', 'D']  # Круг, квадрат, ромб
colors = ['red', 'green', 'purple']
handles = [line_dB]

for i, idx in enumerate(indices):
    freq = data['freq'].iloc[idx]
    s11_db = data['S11_dB'].iloc[idx]
    scatter = plt.scatter(
        freq, s11_db, 
        marker=markers[i], 
        color=colors[i], 
        s=50, 
        edgecolor='black',
        zorder=5,
        label=f'{freq:.1f} МГц: {s11_db:.2f} дБ'
    )
    handles.append(scatter)

# Настройки графика
plt.xlabel('Частота, МГц', fontsize=12)
plt.ylabel('Коэффициент отражения, дБ', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(handles=handles)
plt.tight_layout()
plt.show()

# --- График аргумента S11 ---
plt.figure(figsize=(12, 6))

# Основная кривая
line_ang, = plt.plot(data['freq'], data['angle'], 
                   label='arg(S11)', color='darkorange', linewidth=2)

# Точки на выбранных частотах
handles_ang = [line_ang]

for i, idx in enumerate(indices):
    freq = data['freq'].iloc[idx]
    angle = data['angle'].iloc[idx]
    scatter = plt.scatter(
        freq, angle, 
        marker=markers[i], 
        color=colors[i], 
        s=50, 
        edgecolor='black',
        zorder=5,
        label=f'{freq:.1f} МГц: {angle:.2f}°'
    )
    handles_ang.append(scatter)

# Настройки графика
plt.xlabel('Частота, МГц', fontsize=12)
plt.ylabel('Аргумент коэффициента отражения, градусы', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(handles=handles_ang)
plt.tight_layout()
plt.show()
```

```python
# Загрузка данных
column_names = [
    'freq', 
    'S11_mag', 'S11_ang', 
    'S21_mag', 'S21_ang', 
    'S12_mag', 'S12_ang', 
    'S22_mag', 'S22_ang'
]

data = pd.read_csv('4.4.7 Matrix.csv', skiprows=2, delim_whitespace=True, 
                  header=None, names=column_names)

# Преобразование модулей в децибелы
for param in ['S11', 'S21', 'S12', 'S22']:
    data[f'{param}_dB'] = 20 * np.log10(data[f'{param}_mag'])

# Стилизация графиков
colors = {'S11': 'blue', 'S21': 'green', 'S12': 'red', 'S22': 'purple'}
linestyles = {'mag': '-', 'ang': '--'}

# --- Графики модулей в дБ ---
plt.figure(figsize=(14, 8))
for param in ['S11', 'S21', 'S12', 'S22']:
    plt.plot(data['freq'], data[f'{param}_dB'], 
            label=f'|{param}| (дБ)', 
            color=colors[param], 
            linewidth=2)

plt.xlabel('Частота, МГц', fontsize=12)
plt.ylabel('Коэффициент передачи/отражения, дБ', fontsize=12)
plt.title('Модули S-параметров', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()

# --- Графики аргументов ---
plt.figure(figsize=(14, 8))
for param in ['S11', 'S21', 'S12', 'S22']:
    plt.plot(data['freq'], data[f'{param}_ang'], 
            label=f'arg({param})', 
            color=colors[param], 
            linestyle='--',
            linewidth=2)

plt.xlabel('Частота, МГц', fontsize=12)
plt.ylabel('Аргумент, градусы', fontsize=12)
plt.title('Аргументы S-параметров', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()

# --- Отдельные графики для каждого параметра ---
fig, axes = plt.subplots(4, 2, figsize=(18, 20))

# Настройка подписей
titles = {
    'S11': 'Коэффициент отражения S11',
    'S21': 'Коэффициент передачи (прямой) S21',
    'S12': 'Коэффициент передачи (обратный) S12',
    'S22': 'Коэффициент отражения (выход) S22'
}

for idx, param in enumerate(['S11', 'S21', 'S12', 'S22']):
    # График модуля
    axes[idx, 0].plot(data['freq'], data[f'{param}_dB'], 
                     color=colors[param], 
                     linewidth=2)
    axes[idx, 0].set_title(f'{titles[param]} - Модуль', fontsize=12)
    axes[idx, 0].set_ylabel('дБ', fontsize=10)
    axes[idx, 0].grid(True, linestyle='--', alpha=0.7)
    
    # График аргумента
    axes[idx, 1].plot(data['freq'], data[f'{param}_ang'], 
                     color=colors[param], 
                     linestyle='--',
                     linewidth=2)
    axes[idx, 1].set_title(f'{titles[param]} - Аргумент', fontsize=12)
    axes[idx, 1].set_ylabel('Градусы', fontsize=10)
    axes[idx, 1].grid(True, linestyle='--', alpha=0.7)

# Общие настройки
for ax in axes.flatten():
    ax.set_xlabel('Частота, МГц', fontsize=10)

plt.tight_layout()
plt.show()
```

```python

# Загрузка данных
column_names = [
    'freq', 
    'S11_mag', 'S11_ang', 
    'S21_mag', 'S21_ang', 
    'S12_mag', 'S12_ang', 
    'S22_mag', 'S22_ang'
]

data = pd.read_csv('4.4.7 Matrix.csv', skiprows=2, delim_whitespace=True, 
                  header=None, names=column_names)

C = 85e-12  # Ф
alpha = 4

freq = np.linspace(10, 2010, 102)  # 10 МГц – 2 ГГц
# Расчёт Xc и коэффициентов
Xc = 1 / (2j * np.pi * freq * C)
s21 = 2/((2j*np.pi*freq*C + 2)* 10**(-alpha/20))
# Перевод в дБ
s21_db = 20 * np.log10(np.abs(s21))

# Преобразование модуля S21 в децибелы
data['S21_dB'] = 20 * np.log10(data['S21_mag'])

# Находим индекс частоты 595.858586 МГц
target_freq = 595.858586
idx = data[data['freq'] == target_freq].index[0]
s21_db_value = data['S21_dB'].iloc[idx]
s21_ang_value = data['S21_ang'].iloc[idx]

# --- График модуля S21 в дБ с отметкой ---
plt.figure(figsize=(12, 6))
plt.plot(data['freq'], data['S21_dB'], 
        label='|S21| (дБ)', 
        color='green', 
        linewidth=2)
# Отметка на целевой частоте
plt.scatter(target_freq, s21_db_value, 
           color='red', 
           s=100, 
           zorder=5, 
           label=f'{target_freq:.2f} МГц: {s21_db_value:.2f} дБ')

plt.xlabel('Частота, МГц', fontsize=12)
plt.ylabel('Коэффициент передачи, дБ', fontsize=12)
plt.title('Модуль S21', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()

# --- График аргумента S21 ---
plt.figure(figsize=(12, 6))
plt.plot(data['freq'], data['S21_ang'], 
        label='arg(S21)', 
        color='darkorange', 
        linewidth=2)
plt.xlabel('Частота, МГц', fontsize=12)
plt.ylabel('Аргумент, градусы', fontsize=12)
plt.title('Аргумент S21', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```
