```python
import matplotlib.pyplot as plt
import numpy as np

# Данные для первого графика (h от N0)
N0 = [0,	1,	2, 3,	4,	5.5]
h_N0 = [1,	0.75,	0.65,	0.5,	0.3,	0]

# Данные для второго графика (h от Δφ)
delta_phi = [0, 15, 30, 45, 60, 75, 90]
h_delta_phi = [1, 0.971, 0.868, 0.708, 0.502, 0.260, 0]

# Создаем фигуру с двумя субплoтами
plt.figure(figsize=(12, 5))

# Первый график
plt.subplot(1, 2, 1)
plt.plot(N0, h_N0, 'bo-', linewidth=1.5, markersize=6)
plt.xlabel('N0, мВт/Гц', fontsize=10)
plt.ylabel('h(раскрыв), В', fontsize=10)
plt.title('Зависимость h от N0', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

# Второй график
plt.subplot(1, 2, 2)
plt.plot(delta_phi, h_delta_phi, 'ro-', linewidth=1.5, markersize=6)
plt.xlabel('Δφ, град', fontsize=10)
plt.ylabel('h(раскрыв), В', fontsize=10)
plt.title('Зависимость h от Δφ', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

# Настройка общего отображения
plt.tight_layout()
plt.show()
```
