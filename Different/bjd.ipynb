```python
import matplotlib.pyplot as plt

# Данные
R = [10, 25, 50, 100, 200]  # кОм
Ih_R = [50.8, 23.6, 12.5, 6.41, 3.25]  # мА
C = [0, 5, 10, 15, 20]  # мкФ
Ih_C = [50.8, 88.9, 13.5, 166, 183]  # мА

# Создаём фигуру
plt.figure(figsize=(12, 6))
plt.suptitle(
    "Зависимость Ih от сопротивления и ёмкости\nв трёхфазной сети с изолированной нейтралью", 
    fontsize=14, 
    y=1.02  # Смещение заголовка вверх
)

# График для сопротивления
plt.subplot(1, 2, 1)
plt.plot(R, Ih_R, 'o', color='blue', markersize=8)
plt.title("При C = 0 мкФ", fontsize=12)  # Упрощённый заголовок
plt.xlabel('Сопротивление, кОм')
plt.ylabel('Ток Ih, мА')
plt.grid(True)

# График для ёмкости
plt.subplot(1, 2, 2)
plt.plot(C, Ih_C, 'o', color='red', markersize=8)
plt.title("При R = 10 кОм", fontsize=12)  # Упрощённый заголовок
plt.xlabel('Ёмкость, мкФ')
plt.ylabel('Ток Ih, мА')
plt.grid(True)

# Настройка отступов
plt.tight_layout()
plt.subplots_adjust(top=0.85)  # Место для общего заголовка

plt.show()
```

```python
import numpy as np

# Данные для сети с глухозаземлённой нейтралью
R = [10, 25, 50, 100, 200]  # кОм
Ih_R = [220, 220, 220, 220, 220]  # мА (C = 0 мкФ)

C = [0, 5, 10, 15, 20]  # мкФ
Ih_C = [220, 220, 220, 220, 220]  # мА (R = 10 кОм)

new_xticks = np.arange(0, 220, 20)

# Настройка графиков
plt.figure(figsize=(12, 6))
plt.suptitle(
    "Зависимость Ih от сопротивления и ёмкости\nв сети с глухозаземлённой нейтралью", 
    fontsize=14, 
    y=1.02
)

# График зависимости от сопротивления
plt.subplot(1, 2, 1)
plt.plot(R, Ih_R, 'o', color='green', markersize=8)
plt.title("При C = 0 мкФ", fontsize=12)
plt.xlabel('Сопротивление, кОм')
plt.xticks(new_xticks)
plt.ylabel('Ток Ih, мА')
plt.grid(True)

# График зависимости от ёмкости
plt.subplot(1, 2, 2)
plt.plot(C, Ih_C, 'o', color='purple', markersize=8)
plt.title("При R = 10 кОм", fontsize=12)
plt.xlabel('Ёмкость, мкФ')
plt.ylabel('Ток Ih, мА')
plt.grid(True)

plt.tight_layout()
plt.subplots_adjust(top=0.85)
plt.show()
```
