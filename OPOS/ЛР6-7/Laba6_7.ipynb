```python
import math
# Исходные данные из таблиц
f0 = 84e6  # Гц
L_k = 0.2e-6  # Гн
Pi_k = 3e6  # Гц
Pi_ke = 8.5e6  # Гц
R_A = 50  # Ом
k_sv = 0.3

# Параметры из таблицы 3
g11 = 1.191e-3  # См
C11 = 33.612e-12  # Ф
R_d1 = 47.79e3  # Ом
R_d2 = 16.5e3  # Ом

# Параметры из таблицы 4
g22 = 19.455e-6  # См
C22 = 3.785e-12  # Ф
Y21_abs = 58.133e-3  # См
R_n = 477  # Ом

# Расчет C_кэ
C_ke = 1 / ((2 * math.pi * f0)**2 * L_k)
print(f"1. C_кэ = {C_ke:.2e} Ф = {C_ke*1e12:.2f} пФ")

# Расчет g_k
g_k = 2 * math.pi * Pi_k * C_ke
print(f"2. g_k = {g_k:.2e} См")

# Расчет R_k
R_k = 1 / g_k
print(f"3. R_k = {R_k:.2f} Ом")

# Расчет g_A
g_A = 1 / R_A
print(f"4. g_A = {g_A:.4f} См")

# Расчет γ
gamma = Pi_ke / Pi_k
print(f"5. γ = {gamma:.4f}")

# Расчет m1_1
m1_1 = math.sqrt((gamma * g_k) / (2 * g_A))
print(f"6. m1(1) = {m1_1:.4f}")

# Расчет L_sv
L_sv = (m1_1 / k_sv)**2 * L_k
print(f"7. L_св = {L_sv*1e6:.4f} мкГн")

# Расчет g_n_VTs
g_n_VTs = g11 + 1/R_d1 + 1/R_d2
print(f"8. g_н ВЦ = {g_n_VTs:.2e} См")

# Расчет m2_1
m2_1 = math.sqrt(((gamma - 2) * g_k) / (2 * g_n_VTs))
print(f"9. m2(1) = {m2_1:.4f}")

# Расчет C1 и C2
C1 = C_ke / (1 - m2_1)
C2 = C_ke / m2_1 - C11
print(f"10. C1 = {C1*1e12:.2f} пФ")
print(f"    C2 = {C2*1e12:.2f} пФ")

# Расчет K0s
K0s = m2_1 / (2 * m1_1)
print(f"11. K0с = {K0s:.4f}")

# Расчет параметров УРЧ
g_n = 1 / R_n
print(f"1. g_н УРЧ = {g_n:.4f} См")

# Расчет m2_2
m2_2 = math.sqrt(((gamma - 1)*g_k - g22) / g_n)
print(f"2. m2(2) = {m2_2:.4f}")

# Расчет C1U и C2U
C1U = (C_ke - C22) / (1 - m2_2)
C2U = (C_ke - C22) / m2_2
print(f"3. C1У = {C1U*1e12:.2f} пФ")
print(f"   C2У = {C2U*1e12:.2f} пФ")

# Расчет K0U
K0U = (1 * m2_2 * Y21_abs) / (gamma * g_k)
print(f"4. K0У = {K0U:.2f}")
```
