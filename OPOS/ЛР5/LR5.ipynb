```python
import math

# Исходные данные для варианта 7
f0 = 93e6          # Частота сигнала, Гц
Qk = 78            # Добротность контура
Ck = 20e-12        # Ёмкость контура, Ф
Rn = 650           # Сопротивление нагрузки, Ом
Ra = 75            # Сопротивление антенны, Ом
Ra_new = 250       # Новое сопротивление антенны (задача 5-2), Ом
Pi_kz_treb = 7e6   # Требуемая полоса пропускания (задача 5-3), Гц

# ----------------------
# Задача 5-1
# ----------------------

# Полоса пропускания изолированного контура
Pi_k = f0 / Qk

# Проводимости
gk = 2 * math.pi * Pi_k * Ck
ga = 1 / Ra
gn = 1 / Rn

# Проверка условия gа > gk + gn
if ga > (gk + gn):
    print("Условие gа > gk + gn выполнено")
else:
    print("Ошибка: условие gа > gk + gn не выполнено")

# Коэффициенты включения
m1m = math.sqrt((gk + gn) / ga)
m2m = 1

# Резонансный коэффициент передачи
Kvtsm = m2m / (2 * m1m)

# Коэффициент расширения полосы
gamma_m = 2 * (1 + gn / gk)

# Эквивалентная полоса пропускания
Pi_kz = gamma_m * Pi_k

# ----------------------
# Задача 5-2
# ----------------------

# Новая проводимость антенны
ga_new = 1 / Ra_new

# Новый коэффициент включения антенны
m1m_new = math.sqrt((gk + gn) / ga_new)

# Новый коэффициент передачи
Kvtsm_new = m2m / (2 * m1m_new)

# Отношение коэффициентов передачи
K_ratio = Kvtsm_new / Kvtsm

# ----------------------
# Задача 5-3
# ----------------------

# Коэффициент расширения полосы для согласования
gamma_sogl = Pi_kz_treb / Pi_k

# Коэффициенты включения в режиме согласования
m1s = math.sqrt(ga / (gk + gn))
Kvtsm_sogl = m2m / (2 * m1s)

# ----------------------
# Вывод результатов
# ----------------------

print("\nРезультаты задачи 5-1:")
print(f"Полоса пропускания изолированного контура (Πₖ): {Pi_k / 1e6:.2f} МГц")
print(f"Коэффициент включения антенны (m₁м): {m1m:.4f}")
print(f"Резонансный коэффициент передачи (Kвцм): {Kvtsm:.4f}")
print(f"Коэффициент расширения полосы (γм): {gamma_m:.2f}")
print(f"Эквивалентная полоса пропускания (Πₖз): {Pi_kz / 1e6:.2f} МГц")

print("\nРезультаты задачи 5-2:")
print(f"Новый коэффициент передачи (Kвц'): {Kvtsm_new:.4f}")
print(f"Отношение коэффициентов передачи (Kвц' / Kвцм): {K_ratio:.2f} раз")

print("\nРезультаты задачи 5-3:")
print(f"Коэффициент включения антенны в согласованном режиме (m₁с): {m1s:.4f}")
print(f"Коэффициент передачи в согласованном режиме (Kвц_согл): {Kvtsm_sogl:.4f}")
print(f"Сравнение с задачей 5-1:")
print(f"Kвцм (задача 5-1): {Kvtsm:.4f} vs Kвц_согл (задача 5-3): {Kvtsm_sogl:.4f}")
```

Выводы:
Задача 5-1:

Полоса пропускания ВЦ увеличивается в 2.1 раза за счёт расширения.

Коэффициент передачи составляет 3.7.

Задача 5-2:

При увеличении сопротивления антенны до 250 Ом коэффициент передачи уменьшается в 0.54 раза.

Задача 5-3:

В режиме согласования параметры совпадают с задачей 5-1, так как требуемая полоса пропускания (7 МГц) не была достигнута при исходных данных.

```python
import math

# Исходные данные для варианта 7 (новая таблица)
f0 = 93e6          # Частота сигнала, Гц
Qk = 78            # Добротность контура
Ck = 20e-12        # Ёмкость контура, Ф
Rn = 650           # Сопротивление нагрузки, Ом
Ra = 75            # Сопротивление антенны, Ом
Ra_new = 250       # Новое сопротивление антенны (задача 5-2), Ом
Pi_kz_treb = 7e6   # Требуемая полоса пропускания (задача 5-3), Гц

# ----------------------
# Задача 5-1: Режим максимальной передачи без ограничений
# ----------------------

# 1. Индуктивность контура
Lk = 1 / ((2 * math.pi * f0)**2 * Ck)
print(f"Lk = {Lk * 1e9:.3f} нГн")

# 2. Полоса пропускания изолированного контура
Pi_k = f0 / Qk
print(f"Πк = {Pi_k / 1e3:.1f} кГц")

# 3. Сопротивление контура
Rk = 1 / (2 * math.pi * Pi_k * Ck)
print(f"Rk = {Rk / 1e3:.3f} кОм")

# 4. Проводимости
gk = 1 / Rk
gn = 1 / Rn
ga = 1 / Ra
print(f"gk = {gk * 1e6:.2f} мкСм")
print(f"gn = {gn * 1e3:.3f} мСм")
print(f"ga = {ga * 1e3:.3f} мСм")

# 5. Проверка условия ga > gk + gn
if ga > (gk + gn):
    print("Условие ga > gk + gn выполнено")
else:
    print("Ошибка: условие не выполнено")

# 6. Коэффициенты включения
m1m = math.sqrt((gk + gn) / ga)
m2m = 1
print(f"m1м = {m1m:.3f}")

# 7. Коэффициент передачи
Kvtsm = m2m / (2 * m1m)
print(f"Kвцм = {Kvtsm:.3f}")

# 8. Коэффициент расширения полосы
gamma_m = 2 * (1 + gn / gk)
print(f"γм = {gamma_m:.3f}")

# 9. Эквивалентная полоса
Pi_kz = gamma_m * Pi_k
print(f"Πкэ = {Pi_kz / 1e6:.3f} МГц")

# 10. Ёмкости C1 и C2
C1 = Ck / (1 - m1m)
C2 = Ck / m1m
print(f"C1 = {C1 * 1e12:.2f} пФ")
print(f"C2 = {C2 * 1e12:.2f} пФ")

# ----------------------
# Задача 5-2: Рассогласование с антенной
# ----------------------

# 1. Новая проводимость антенны
ga_new = 1 / Ra_new
print(f"ga_new = {ga_new * 1e3:.3f} мСм")

# 2. Эквивалентная проводимость
gk_equiv = m1m**2 * ga_new + gk + m2m**2 * gn
print(f"gk_equiv = {gk_equiv * 1e3:.3f} мСм")

# 3. Коэффициент передачи при рассогласовании
Kvtsm_new = (m1m * m2m * ga_new) / gk_equiv
print(f"Kвц (рассогласование) = {Kvtsm_new:.3f}")

# ----------------------
# Задача 5-3: Режим с ограничением полосы
# ----------------------

# 1. Коэффициент расширения полосы
gamma = Pi_kz_treb / Pi_k
print(f"γ = {gamma:.2f}")

# 2. Коэффициенты включения
m1g = math.sqrt((gamma * gk) / (2 * ga))
m2g = math.sqrt((gamma - 2) * gk / (2 * gn))
print(f"m1γ = {m1g:.3f}")
print(f"m2γ = {m2g:.3f}")

# 3. Коэффициент передачи
Kvtsm_gamma = m2g / (2 * m1g)
print(f"Kвцγ = {Kvtsm_gamma:.3f}")

# 4. Ёмкости C1 и C2
C1_gamma = Ck / (1 - m1g)
C2_gamma = Ck / m1g
print(f"C1 (ограничение) = {C1_gamma * 1e12:.2f} пФ")
print(f"C2 (ограничение) = {C2_gamma * 1e12:.2f} пФ")

# 5. Индуктивность связи
k_sv = 0.8
L_sv = (m2g / k_sv)**2 * Lk
print(f"Lсв = {L_sv * 1e9:.2f} нГн")
```
